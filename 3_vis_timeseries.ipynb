# Load libraries & authenticate Earth Engine

```python
import ee
import geemap

ee.Authenticate()
ee.Initialize()

ee_project = '<YOUR_EE_PROJECT_NAME>'
START_Y = 2024
END_Y = 2025
roi_path = '<your_roi_path>'
```

# Visualise metadata collection

```python
path = f'projects/{ee_project}/assets/snowMapper/{roi_path}/meta_{roi_path}_{START_Y}_{END_Y}'

variable = {
    'sc_obs': True,
    'sc_modis': False,
    'sc_probab': False,
    'flag_all': False,
    't2mHres': False,
    'precipHres': False,
    'hd_precip': False,
    'cd_precip': False,
    'hd': False,
    'cd': False
}

dt = 2

Map = geemap.Map()
ic = ee.ImageCollection(path).sort('system:time_start')
# Map.addLayer(ic.first().select('elev'), {'min': 0, 'max': 4000, 'palette': ['white', 'black']}, 'Elevation', False)
 
if variable['sc_obs']:
    Map.add_time_slider(ic.map(lambda img: img.select('sc_obs')), 
                        {'min': 0, 'max': 2, 'palette': ['yellow', 'purple', 'pink']}, 
                        time_interval=dt)
    
if variable['sc_modis']:
    Map.add_time_slider(ic.map(lambda img: img.select('sc_modis')), 
                        {'min': 0, 'max': 2, 'palette': ['yellow', 'purple', 'pink']}, 
                        time_interval=dt)

if variable['sc_probab']:
    Map.add_time_slider(ic.map(lambda img: img.select('sc_probab')), 
                        {'min': 0, 'max': 1, 'palette': ['yellow', 'purple']}, 
                        time_interval=dt)

if variable['flag_all']:
    Map.add_time_slider(ic.map(lambda img: img.select('flag_all')), 
                        {'min': 0, 'max': 1, 'palette': ['green', 'red']}, 
                        time_interval=dt)
    
if variable['t2mHres']:
    Map.add_time_slider(ic.map(lambda img: img.select('t2mHres')), 
                        {'min': -5, 'max': 20, 'palette': ['blue', 'white', 'red']}, 
                        time_interval=dt)
    
if variable['precipHres']:
    Map.add_time_slider(ic.map(lambda img: img.select('precipHres')), 
                        {'min': 0, 'max': 25, 'palette': ['yellow', 'orange', 'purple', 'red']},
                        time_interval=dt)
    
if variable['hd_precip']:
    Map.add_time_slider(ic.map(lambda img: img.select('hd_precip')), 
                        {'min': 0, 'max': 25, 'palette': ['yellow', 'orange', 'purple', 'red']},
                        time_interval=dt)

if variable['cd_precip']:
    Map.add_time_slider(ic.map(lambda img: img.select('cd_precip')), 
                        {'min': 0, 'max': 25, 'palette': ['yellow', 'orange', 'purple', 'red']},
                        time_interval=dt)

if variable['hd']:
    Map.add_time_slider(ic.map(lambda img: img.select('hd').updateMask(img.select('hd').gt(0))), 
                        {'min': 0, 'max': 10, 'palette': ['blue', 'white', 'red']}, 
                        time_interval=dt)

if variable['cd']:
    Map.add_time_slider(ic.map(lambda img: img.select('cd').updateMask(img.select('cd').gt(0))), 
                        {'min': 0, 'max': 15, 'palette': ['red', 'white', 'blue']}, 
                        time_interval=dt)

Map.centerObject(ic.first())
Map
```

# Visualise multiyear snow cover probabilities

```python
path = f'projects/{ee_project}/assets/snowMapper/{roi_path}/sc_probab_{roi_path}_{START_Y}_{END_Y}'

dt = 2

Map = geemap.Map()

def add_time_start(img):
    month = ee.Number(img.get('month')).format('%02d')  # Zero-padded month
    date_str = ee.String('1984-').cat(month).cat('-01')
    date = ee.Date(date_str)
    return img.set('system:time_start', date.millis())

ic = ee.ImageCollection(path).map(add_time_start).sort('system:time_start')

 # .set({'month': collection.first().date().get('month')})
Map.add_time_slider(ic.map(lambda img: img.select('sc_probab').reproject(crs="EPSG:2100", scale=100)), 
                    {'min': 0, 'max': 1, 'palette': ['yellow', 'purple']}, 
                    time_interval=dt)
    
Map.centerObject(ic.first())
Map
```
